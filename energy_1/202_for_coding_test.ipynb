```python
# 조기 반환  : 조기 반환 뒤의 예외처리 미실행 (시간복잡도↓)
def total_price(quantity, price):
    total = quantity * price
    if total > 100:
        return total * 0.9      # <- 조기 반환
    return total

print(total_price(4,50))
```

```python
# 보호 구문 : 예외처리를 먼저 처리하는 방법
def calculate_average(numbers) :
    if numbers is None:                     # <- 예외처리 1
        return None
    
    if not isinstance(numbers, list) :      # <- 예외처리 2
        return None
    
    if len(numbers):
        return None                         # <- 예외처리 3

    total = sum(numbers)
    average = total / len(numbers)
    return average
```

```python
# 합성함수 사용 : 2+개 함수 활용한 사용자 함수 생성  -> 코드 관리와 사용이 쉬워짐
def add_three(x):
    return x+3

def square(x):
    return x*x

composed_function = lambda x: square(add_three(x))          # 합성함수 (squre x add_three)
print(composed_function(3))             # (3+3)*(3+3) = 6*6 = 36
```

```python
# Q02 : 배열 제어하기
def my_solution (arr):
    arr.sort(reverse=True)
    result = [arr[0]]
    for idx, num in enumerate(arr):
        if idx>0:
            if arr[idx-1] != arr[idx]:
                result.append(num)
    return result

def book_solution(lst):
  unique_lst = list(set(lst)) # ➊ 중복값 제거
  unique_lst.sort(reverse=True) # ➋ 내림차순 정렬 
  return unique_lst

print(my_solution([4,2,2,1,3,4]))
print(my_solution([2,1,1,3,2,5,4]))
print(book_solution([4,2,2,1,3,4]))
print(book_solution([2,1,1,3,2,5,4]))
```
